```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 4 * np.pi, 500)
wave1 = np.sin(x)
wave2 = np.sin(2 * x)
interference = wave1 + wave2

plt.figure(figsize=(8, 4))
plt.plot(x, wave1, '--', label='Base Wave (sin x)')
plt.plot(x, interference, label='Superposition Wave (sin x + sin 2x)', linewidth=2)
plt.title('Ananth R&D Lab — Day 1 Wave Superposition Test')
plt.xlabel('Space (x)')
plt.ylabel('Amplitude')
plt.legend()
plt.grid(True)
plt.show()
```
