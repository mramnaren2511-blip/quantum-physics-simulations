```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Define the 1D Space
x = np.linspace(-5, 5, 1000)

# 2. Create an unnormalized wave function (1D Gaussian)
psi_unnormalized = np.exp(-x**2)
prob_density_unnormalized = np.abs(psi_unnormalized)**2

# 3. Calculate Initial Area (Probability)
# FIXED: np.trapz badulu np.trapezoid vaadali in latest NumPy
initial_area = np.trapezoid(prob_density_unnormalized, x)
print(f"Area Before Normalization: {initial_area:.4f}")

# 4. NORMALIZE THE WAVE
A = 1 / np.sqrt(initial_area)
print(f"Normalization Constant (A): {A:.4f}")

psi_normalized = A * psi_unnormalized
prob_density_normalized = np.abs(psi_normalized)**2

# 5. Verify the Final Area
final_area = np.trapezoid(prob_density_normalized, x)
print(f"Area After Normalization: {final_area:.4f} (Rule of Existence Confirmed)")

# 6. Visualize
plt.figure(figsize=(10, 5))
plt.plot(x, prob_density_unnormalized, label=r'Unnormalized $|\Psi|^2$', linestyle='--', color='red')
plt.plot(x, prob_density_normalized, label=r'Normalized $|\Psi|^2$ (Area = 1)', color='blue')
plt.fill_between(x, prob_density_normalized, color='blue', alpha=0.2)
plt.title('Griffiths 1.3: 1D Quantum Wave Normalization')
plt.xlabel('Position (x)')
plt.ylabel(r'Probability Density $|\Psi|^2$')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```
